```python
# Simple Internal Banking RAG Prototype
# This notebook demonstrates a basic Retrieval-Augmented Generation pipeline
# for querying internal banking documents.

# Import libraries
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

# --------------------------------------------------
# STEP 1: Simulated internal banking documents
# --------------------------------------------------

documents = [
"""
Loan approvals require credit risk assessment.
Customer income verification and identity verification are mandatory.
All lending operations must follow internal compliance guidelines.
""",

"""
Customer onboarding requires identity verification and AML checks.
The bank must comply with anti-money laundering regulations.
Suspicious transactions must be reported to the compliance department.
""",

"""
Mortgage applications require evaluation of customer credit score,
loan-to-value ratio, and financial stability.
Risk management policies must be followed before approval.
"""
]

# --------------------------------------------------
# STEP 2: Chunk documents
# --------------------------------------------------

text_splitter = CharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = []

for doc in documents:
    chunks.extend(text_splitter.split_text(doc))

print("Document chunks created:", len(chunks))

# --------------------------------------------------
# STEP 3: Create embeddings
# --------------------------------------------------

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# --------------------------------------------------
# STEP 4: Create vector database
# --------------------------------------------------

vector_db = Chroma.from_texts(
    texts=chunks,
    embedding=embedding_model
)

print("Vector database created.")

# --------------------------------------------------
# STEP 5: Query the system
# --------------------------------------------------

query = "What checks are required for loan approval?"

results = vector_db.similarity_search(query, k=2)

print("\nQuery:", query)
print("\nRelevant internal documents:\n")

for r in results:
    print(r.page_content)
    print("---")
```
